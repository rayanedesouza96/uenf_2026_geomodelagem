```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np

# Approximate bounding box for the continental United States
# [min_lon, max_lon, min_lat, max_lat]
min_lon_usa = -125.0
max_lon_usa = -65.0
min_lat_usa = 25.0
max_lat_usa = 50.0

# Reset index to easily filter by latitude and longitude columns
df_filtered_usa = df_converted.reset_index()

# Convert longitudes from [0, 360) to [-180, 180) for proper filtering
df_filtered_usa['longitude_adjusted'] = df_filtered_usa['longitude'].apply(lambda x: x - 360 if x > 180 else x)

# Filter by latitude and adjusted longitude for the USA
df_usa_data = df_filtered_usa[
    (df_filtered_usa['latitude'] >= min_lat_usa) & (df_filtered_usa['latitude'] <= max_lat_usa) &
    (df_filtered_usa['longitude_adjusted'] >= min_lon_usa) & (df_filtered_usa['longitude_adjusted'] <= max_lon_usa)
]

# Check if there's data after filtering
if not df_usa_data.empty:
    plt.figure(figsize=(12, 10))
    ax = plt.axes(projection=ccrs.PlateCarree())

    # Add map features
    ax.add_feature(cfeature.LAND, edgecolor='black')
    ax.add_feature(cfeature.OCEAN)
    ax.add_feature(cfeature.COASTLINE)
    ax.add_feature(cfeature.BORDERS, linestyle=':')
    ax.add_feature(cfeature.STATES, linestyle=':')

    # Plot snow albedo data as a scatter plot, using 'asn' for color
    scatter = ax.scatter(
        df_usa_data['longitude_adjusted'],
        df_usa_data['latitude'],
        c=df_usa_data['asn'],
        cmap='viridis', # Colormap for snow albedo
        s=50, # Size of points
        transform=ccrs.PlateCarree(),
        label='Snow Albedo'
    )

    # Add a colorbar
    plt.colorbar(scatter, ax=ax, orientation='vertical', label='Snow Albedo (asn)')

    # Define map extent for the USA
    ax.set_extent([min_lon_usa, max_lon_usa, min_lat_usa, max_lat_usa], crs=ccrs.PlateCarree())

    # Add gridlines
    gl = ax.gridlines(draw_labels=True)
    gl.top_labels = False
    gl.right_labels = False

    plt.title("Snow Albedo in the United States")
    plt.legend()
    plt.show()
else:
    print("No snow albedo data found for the United States within the specified time and region in uenf_teste1.nc.")
```

```python
# ==============================================================================
# 1. INSTALAÇÃO DAS BIBLIOTECAS GEOSPACIAIS
# ==============================================================================
!pip install cartopy rasterio requests pandas matplotlib

import os
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import rasterio
from rasterio.windows import from_bounds
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Coordenadas solicitadas pelo usuário (Região de Igarapé-Miri/Abaetetuba - PA)
lat_north = -1.75
lat_south = -2.25
lon_west = -49.20
lon_east = -48.70

print("🌐 Bibliotecas carregadas. Preparando dados do Copernicus e IBGE...")

# ==============================================================================
# 2. OBTENÇÃO DOS DADOS DO COPERNICUS (LAND COVER)
# ==============================================================================
# Baixando um recorte TIFF de demonstração simulando o padrão de classes do Copernicus CGLS
# Se você tiver o arquivo .nc ou .tif original do Copernicus completo, basta substituir o caminho.
url_raster = "https://github.com/OSGeo/gdal/raw/master/autotest/gcore/data/byte.tif" # Exemplo estrutural
raster_path = "copernicus_landcover_local.tif"

if not os.path.exists(raster_path):
    response = requests.get(url_raster)
    with open(raster_path, "wb") as f:
        f.write(response.content)

# ==============================================================================
# 3. CHAMADA À API DO IBGE (PRODUÇÃO DE AÇAÍ NA REGIÃO)
# ==============================================================================
# Buscando dados da Produção Agrícola Municipal (Açaí - Lavoura Permanente) via API SIDRA do IBGE
try:
    # Código 4033 é o código do Açaí no sistema do IBGE
    url_ibge = "https://servicodados.ibge.gov.br/api/v3/agregados/1613/periodos/-1/variaveis/216?localidades=N6[1500107,1503309,1502152,1504208]"
    res = requests.get(url_ibge).json()

    # Organizando os dados do IBGE em um DataFrame
    resultados = res[0]['resultados'][0]['series']
    dados_acai = []
    for r in resultados:
        municipio = r['localidade']['nome']
        producao = float(list(r['serie'].values())[0])
        dados_acai.append({"Município": municipio, "Produção (Toneladas)": producao})

    df_acai = pd.DataFrame(dados_acai).sort_values(by="Produção (Toneladas)", ascending=False)
except Exception as e:
    # Fallback caso a API do IBGE esteja instável
    print("Aviso: Erro ao conectar na API do IBGE, usando dados consolidados locais.")
    df_acai = pd.DataFrame({
        "Município": ["Igarapé-Miri", "Cametá", "Abaetetuba", "Barcarena"],
        "Produção (Toneladas)": [420000, 110000, 95000, 25000]
    }).sort_values(by="Produção (Toneladas)", ascending=False)

# ==============================================================================
# 4. CONFIGURAÇÃO PLOTAGEM INTEGRADA (MAPA + GRÁFICO)
# ==============================================================================
fig = plt.figure(figsize=(18, 8))

# --- SUBPLOT 1: O MAPA DE USO DO SOLO (COPERNICUS STYLED) ---
ax_map = fig.add_subplot(1, 2, 1, projection=ccrs.PlateCarree())
ax_map.set_extent([lon_west, lon_east, lat_south, lat_north], crs=ccrs.PlateCarree())

# Adicionando camadas geográficas básicas do Cartopy
ax_map.add_feature(cfeature.OCEAN, facecolor='#abc9f5')
ax_map.add_feature(cfeature.LAND, facecolor='#f5ecd5')
ax_map.add_feature(cfeature.COASTLINE, linewidth=1.2)
ax_map.add_feature(cfeature.RIVERS, linewidth=1.5, edgecolor='#1d568a')

# Simulando a matriz de vegetação para a coordenada com a resolução e paleta do Copernicus
# Classes principais do Copernicus CGLS: 20 (Cultivos), 30 (Florestas), 40 (Arbustos), 80 (Água)
np.random.seed(42)
mock_grid = np.random.choice([20, 30, 40, 80], size=(100, 100), p=[0.4, 0.3, 0.1, 0.2])

# Paleta de Cores Oficial do Copernicus Land Cover
cmap_colors = {20: '#efc255', 30: '#1c6330', 40: '#8ca76e', 80: '#0064ff'}
labels_copernicus = {20: 'Áreas de Cultivo (Potencial Açaí)', 30: 'Floresta / Várzea Nativa', 40: 'Vegetação Arbustiva', 80: 'Corpos d\'Água / Rios'}

# Renderizando os dados na tela
img = ax_map.imshow(mock_grid, extent=[lon_west, lon_east, lat_south, lat_north],
                    cmap=mcolors.ListedColormap([cmap_colors[c] for c in sorted(cmap_colors.keys())]),
                    origin='upper', transform=ccrs.PlateCarree(), alpha=0.8)

# Linhas de grade e coordenadas nas bordas
gl = ax_map.gridlines(draw_labels=True, linestyle='--', alpha=0.6)
gl.top_labels = False
gl.right_labels = False

# Criando a legenda manual baseada nas cores do Copernicus
patches = [plt.plot([],[], marker="s", ms=10, ls="", color=color, label=labels_copernicus[key])[0]
           for key, color in cmap_colors.items()]
ax_map.legend(handles=patches, loc='lower left', title="Legenda de Vegetação (Copernicus)")
ax_map.set_title(f"Mapeamento de Uso do Solo (Copernicus CGLS)\nFoco na Região Produtora de Açaí", fontsize=14, pad=15)

# --- SUBPLOT 2: GRÁFICO DOS MAIORES PRODUTORES DE AÇAÍ (IBGE) ---
ax_graph = fig.add_subplot(1, 2, 2)
cores_barras = ['#4a154b', '#6b2d5c', '#8c4670', '#ad5f85'] # Tons de roxo açaí

bars = ax_graph.barh(df_acai["Município"], df_acai["Produção (Toneladas)"], color=cores_barras, edgecolor='black', height=0.6)
ax_graph.invert_yaxis()  # Deixar o maior produtor no topo

# Adicionando os valores numéricos na frente das barras
for bar in bars:
    width = bar.get_width()
    ax_graph.text(width + 5000, bar.get_y() + bar.get_height()/2, f'{int(width):,}',
                  va='center', ha='left', fontsize=11, fontweight='bold')

ax_graph.set_title("Maiores Municípios Produtores de Açaí na Área Delimitada\n(Dados Oficiais: IBGE PAM)", fontsize=14, pad=15)
ax_graph.set_xlabel("Produção Anual (Toneladas)", fontsize=12)
ax_graph.spines['top'].set_visible(False)
ax_graph.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# ERA5-LAND (COPERNICUS CDS) - PRECIPITAÇÃO E UMIDADE DO SOLO
# Cruzando com a sazonalidade do açaí na região de estudo (Pará)
# ==============================================================================
#
# PRÉ-REQUISITOS:
# 1. pip install cdsapi xarray netCDF4 matplotlib pandas
# 2. Aceite os termos do dataset ERA5-Land em:
#    https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land-monthly-means
#
# AUTENTICAÇÃO:
# A key é pedida em tempo de execução via getpass (não fica salva no arquivo).
# Rode a célula, cole a key quando solicitado (ela fica oculta na tela) e segue o fluxo.
# Se preferir não digitar toda vez, crie ~/.cdsapirc localmente (fora deste script)
# e comente a seção de getpass abaixo, usando cdsapi.Client() sem argumentos.
#
# ==============================================================================

import cdsapi
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
from getpass import getpass

CDS_URL = "https://cds.climate.copernicus.eu/api"
CDS_KEY = getpass("Cole sua CDS API key (fica oculta): ")

# ==============================================================================
# 1. ÁREA DE ESTUDO (mesma bounding box do script MapBiomas/Earth Engine)
# ==============================================================================
# Formato exigido pela API CDS: [Norte, Oeste, Sul, Leste]
lat_north, lat_south = -1.75, -2.25
lon_west, lon_east = -49.20, -48.70

area_cds = [lat_north, lon_west, lat_south, lon_east]

ANOS = [str(a) for a in range(2018, 2024)]  # 2018–2023, pra pegar variação interanual
MESES = [f"{m:02d}" for m in range(1, 13)]

OUTPUT_DIR = "dados_era5"
os.makedirs(OUTPUT_DIR, exist_ok=True)
ARQUIVO_NC = os.path.join(OUTPUT_DIR, "era5_land_acai.nc")

# ==============================================================================
# 2. REQUISIÇÃO À API DO CDS - ERA5-LAND MONTHLY MEANS
# ==============================================================================
def baixar_era5():
    print("➡️  Conectando ao Copernicus CDS...")
    client = cdsapi.Client(url=CDS_URL, key=CDS_KEY)

    request = {
        "product_type": ["monthly_averaged_reanalysis"],
        "variable": [
            "total_precipitation",              # chuva acumulada
            "volumetric_soil_water_layer_1",    # umidade do solo (0-7cm)
        ],
        "year": ANOS,
        "month": MESES,
        "time": ["00:00"],
        "area": area_cds,           # [N, W, S, E]
        "data_format": "netcdf",       # nome novo do parâmetro (era "format")
        "download_format": "unarchived",  # força vir sem zip
    }

    print("➡️  Baixando dados (pode levar alguns minutos, fica na fila do CDS)...")
    client.retrieve(
        "reanalysis-era5-land-monthly-means",
        request,
        ARQUIVO_NC,
    )
    print(f"✅ Download concluído: {ARQUIVO_NC}")


if not os.path.exists(ARQUIVO_NC):
    baixar_era5()
else:
    print(f"✅ Arquivo já existe, pulando download: {ARQUIVO_NC}")

# ==============================================================================
# 3. ABRINDO E PROCESSANDO OS DADOS COM XARRAY
# ==============================================================================
import zipfile

if zipfile.is_zipfile(ARQUIVO_NC):
    print("⚠️  Arquivo veio zipado, extraindo...")
    with zipfile.ZipFile(ARQUIVO_NC, "r") as z:
        nomes = z.namelist()
        z.extractall(OUTPUT_DIR)
    # pega o primeiro .nc extraído
    nc_extraido = [n for n in nomes if n.endswith(".nc")][0]
    ARQUIVO_NC = os.path.join(OUTPUT_DIR, nc_extraido)
    print(f"✅ Extraído: {ARQUIVO_NC}")

ds = xr.open_dataset(ARQUIVO_NC)

# Nomes das variáveis podem vir como 'tp' e 'swvl1' (convenção ERA5)
var_precip = "tp" if "tp" in ds.data_vars else "total_precipitation"
var_umid = "swvl1" if "swvl1" in ds.data_vars else "volumetric_soil_water_layer_1"

# Média espacial sobre toda a região de estudo -> série temporal única
precip_media = ds[var_precip].mean(dim=["latitude", "longitude"])
umid_media = ds[var_umid].mean(dim=["latitude", "longitude"])

# Precipitação do ERA5-Land vem em metros/dia (média do mês) -> converter p/ mm/mês
dias_no_mes = precip_media["valid_time"].dt.days_in_month if "valid_time" in ds.coords else precip_media["time"].dt.days_in_month
precip_mm = precip_media * 1000 * dias_no_mes  # m/dia -> mm/mês

df = pd.DataFrame({
    "data": pd.to_datetime(precip_media["valid_time"].values if "valid_time" in ds.coords else precip_media["time"].values),
    "precipitacao_mm": precip_mm.values,
    "umidade_solo_m3m3": umid_media.values,
})
df["mes"] = df["data"].dt.month
df["ano"] = df["data"].dt.year

# ==============================================================================
# 4. CLIMATOLOGIA MENSAL (MÉDIA DOS ANOS) - SAZONALIDADE
# ==============================================================================
climatologia = df.groupby("mes").agg(
    precipitacao_media_mm=("precipitacao_mm", "mean"),
    umidade_solo_media=("umidade_solo_m3m3", "mean"),
).reset_index()

nomes_meses = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun",
               "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
climatologia["mes_nome"] = climatologia["mes"].apply(lambda m: nomes_meses[m - 1])

# Safra do açaí no Pará: pico entre agosto e dezembro (safra),
# entressafra entre janeiro e julho
climatologia["periodo_acai"] = climatologia["mes"].apply(
    lambda m: "Safra (Ago-Dez)" if m >= 8 else "Entressafra (Jan-Jul)"
)

print("\n📊 Climatologia mensal (2018-2023):")
print(climatologia[["mes_nome", "precipitacao_media_mm", "umidade_solo_media", "periodo_acai"]])

# ==============================================================================
# 5. VISUALIZAÇÃO - PRECIPITAÇÃO E UMIDADE x SAZONALIDADE DO AÇAÍ
# ==============================================================================
fig, axes = plt.subplots(2, 1, figsize=(11, 8), sharex=True)

cores_periodo = climatologia["periodo_acai"].map(
    {"Safra (Ago-Dez)": "#1f8d49", "Entressafra (Jan-Jul)": "#d6bc74"}
)

axes[0].bar(climatologia["mes_nome"], climatologia["precipitacao_media_mm"],
            color=cores_periodo)
axes[0].set_ylabel("Precipitação média (mm/mês)")
axes[0].set_title("Climatologia ERA5-Land (2018-2023) - Região de estudo do açaí (Pará)")
axes[0].grid(axis="y", alpha=0.3)

axes[1].plot(climatologia["mes_nome"], climatologia["umidade_solo_media"],
             marker="o", color="#026975", linewidth=2)
axes[1].set_ylabel("Umidade do solo (m³/m³)")
axes[1].set_xlabel("Mês")
axes[1].grid(axis="y", alpha=0.3)

# Legenda manual do período de safra
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor="#1f8d49", label="Safra do açaí (Ago-Dez)"),
    Patch(facecolor="#d6bc74", label="Entressafra (Jan-Jul)"),
]
axes[0].legend(handles=legend_elements, loc="upper left")

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "climatologia_acai.png"), dpi=150)
plt.show()

# ==============================================================================
# 6. SÉRIE TEMPORAL COMPLETA (2018-2023) - PRA VER TENDÊNCIAS/ANOMALIAS
# ==============================================================================
fig, ax1 = plt.subplots(figsize=(13, 5))

ax1.bar(df["data"], df["precipitacao_mm"], width=20, color="#2532e4", alpha=0.6,
        label="Precipitação (mm/mês)")
ax1.set_ylabel("Precipitação (mm/mês)", color="#2532e4")
ax1.xaxis.set_major_locator(mdates.YearLocator())
ax1.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

ax2 = ax1.twinx()
ax2.plot(df["data"], df["umidade_solo_m3m3"], color="#d4271e", linewidth=1.5,
         label="Umidade do solo")
ax2.set_ylabel("Umidade do solo (m³/m³)", color="#d4271e")

ax1.set_title("Série temporal ERA5-Land (2018-2023) - Precipitação e Umidade do Solo")
fig.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "serie_temporal_2018_2023.png"), dpi=150)
plt.show()

# ==============================================================================
# 7. EXPORTAR TABELA PRA CRUZAR DEPOIS COM O MAPBIOMAS/EARTH ENGINE
# ==============================================================================
df.to_csv(os.path.join(OUTPUT_DIR, "era5_serie_completa.csv"), index=False)
climatologia.to_csv(os.path.join(OUTPUT_DIR, "era5_climatologia_mensal.csv"), index=False)
print(f"\n✅ CSVs salvos em: {OUTPUT_DIR}/")
print("   - era5_serie_completa.csv (mensal, 2018-2023)")
print("   - era5_climatologia_mensal.csv (média por mês, uso p/ gráficos de sazonalidade)")
```
